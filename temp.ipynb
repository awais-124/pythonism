```python
def findMedianSortedArrays(nums1, nums2):
        merged_array = [*nums1, *nums2]
        merged_array.sort()
        total_length = len(nums1) + len(nums2)
        is_even = total_length % 2 == 0
        median = 0
        half = int(total_length/2)
        print("HALF : ", half)
        if not is_even:
            print("IS NOT EVEN")
            median = merged_array[half]  
        else:
            print("EVEN")
            median = merged_array[half-1] + merged_array[half]
            median = median / 2
            
        return median 
```

```python
def isPalindrome( x: int) -> bool:
        return str(x) == str(x)[::-1]
```
