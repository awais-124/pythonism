```python
# 1. capitalize()
s = "hello world"
print(s.capitalize())  # Output: "Hello world"
# Capitalizes the first letter of the string.

# 2. upper()
print(s.upper())  # Output: "HELLO WORLD"
# Converts all characters to uppercase.

# 3. lower()
print(s.lower())  # Output: "hello world"
# Converts all characters to lowercase.

# 4. title()
print(s.title())  # Output: "Hello World"
# Capitalizes the first letter of each word.

# 5. swapcase()
print(s.swapcase())  # Output: "HELLO WORLD"
# Swaps the case of all letters.

# 6. find(sub[, start[, end]])
print(s.find("world"))  # Output: 6
# Finds the first occurrence of 'sub'. Returns -1 if not found.

# 7. index(sub[, start[, end]])
print(s.index("world"))  # Output: 6
# Similar to `find()`, but raises ValueError if not found.

# 8. replace(old, new[, count])
print(s.replace("world", "Python"))  # Output: "hello Python"
# Replaces occurrences of 'old' with 'new'. Optionally limit replacements with 'count'.

# 9. split(sep=None, maxsplit=-1)
print(s.split())  # Output: ['hello', 'world']
# Splits string by 'sep'. If None, splits by whitespace.

# 10. join(iterable)
words = ["join", "these", "words"]
print(" ".join(words))  # Output: "join these words"
# Joins elements of iterable using the string as a separator.

# 11. strip([chars])
s2 = "  hello  "
print(s2.strip())  # Output: "hello"
# Removes leading/trailing characters (default is whitespace).

# 12. startswith(prefix[, start[, end]])
print(s.startswith("hello"))  # Output: True
# Returns True if the string starts with 'prefix'.

# 13. endswith(suffix[, start[, end]])
print(s.endswith("world"))  # Output: True
# Returns True if the string ends with 'suffix'.

# 14. count(sub[, start[, end]])
print(s.count("l"))  # Output: 3
# Counts occurrences of 'sub' in the string.

# 15. isalpha(), isdigit(), isalnum()
print("hello".isalpha())  # Output: True (all alphabetic)
print("123".isdigit())  # Output: True (all numeric)
print("abc123".isalnum())  # Output: True (all alphanumeric)

# 16. format(*args, **kwargs)
print("Name: {}, Age: {}".format("Alice", 30))  # Output: "Name: Alice, Age: 30"
# Inserts arguments into the string.

# 17. zfill(width)
print("42".zfill(5))  # Output: "00042"
# Pads string with leading zeros to make it 'width' characters long.
```

```python
# 1. append(x)
lst = [1, 2, 3]
lst.append(4)
print(lst)  # Output: [1, 2, 3, 4]
# Adds an element to the end of the list.

# 2. extend(iterable)
lst.extend([5, 6])
print(lst)  # Output: [1, 2, 3, 4, 5, 6]
# Extends the list with elements from the iterable.

# 3. insert(i, x)
lst.insert(2, 99)
print(lst)  # Output: [1, 2, 99, 3, 4, 5, 6]
# Inserts 'x' at index 'i'.

# 4. remove(x)
lst.remove(99)
print(lst)  # Output: [1, 2, 3, 4, 5, 6]
# Removes the first occurrence of 'x'. Raises ValueError if not found.

# 5. pop([i])
lst.pop()
print(lst)  # Output: [1, 2, 3, 4, 5]
# Removes and returns the element at index 'i'. Default is the last element.

# 6. clear()
lst.clear()
print(lst)  # Output: []
# Removes all elements from the list.

# 7. index(x[, start[, end]])
lst = [10, 20, 30, 20]
print(lst.index(20))  # Output: 1
# Returns the first index of 'x'.

# 8. count(x)
print(lst.count(20))  # Output: 2
# Returns the number of occurrences of 'x'.

# 9. reverse()
lst.reverse()
print(lst)  # Output: [20, 30, 20, 10]
# Reverses the list in place.

# 10. sort(key=None, reverse=False)
lst.sort()
print(lst)  # Output: [10, 20, 20, 30]
# Sorts the list in ascending order. Can use custom 'key' or 'reverse' sorting.

# 11. copy()
copy_lst = lst.copy()
print(copy_lst)  # Output: [10, 20, 20, 30]
# Creates a shallow copy of the list.
```

```python
# 1. keys()
d = {'a': 1, 'b': 2}
print(d.keys())  # Output: dict_keys(['a', 'b'])
# Returns a view of keys.

# 2. values()
print(d.values())  # Output: dict_values([1, 2])
# Returns a view of values.

# 3. items()
print(d.items())  # Output: dict_items([('a', 1), ('b', 2)])
# Returns a view of (key, value) pairs.

# 4. get(key[, default])
print(d.get('a'))  # Output: 1
# Returns the value for 'key'. Returns 'default' if key not found.

# 5. pop(key[, default])
print(d.pop('a'))  # Output: 1
print(d)  # Output: {'b': 2}
# Removes the key and returns its value.

# 6. popitem()
print(d.popitem())  # Output: ('b', 2)
print(d)  # Output: {}
# Removes and returns the last (key, value) pair.

# 7. update([other])
d.update({'c': 3})
print(d)  # Output: {'c': 3}
# Updates the dictionary with another dictionary or iterable of (key, value) pairs.

# 8. clear()
d.clear()
print(d)  # Output: {}
# Removes all items.

# 9. copy()
d_copy = {'x': 10}.copy()
print(d_copy)  # Output: {'x': 10}
# Creates a shallow copy.
```
